2. Предложите способ заменить отсутствующие значения из пункта 1 и реализуйте его. Способ нужно аргументировать(сказать почему это достаточно хорошее приближение)

````python
'''
можно посчитать во сколько раз полная площадь квартиры превосходит жилую
и взять медиану от полученных значений, так как она наиболее устойчива к выбросам
'''
k = (df['full_sq'] / df['life_sq']).median()
df = df.copy()
df['life_sq'] = df['life_sq'].fillna(df['full_sq'] / k)
df
````

### 3. Визуализация

Используя библиотеки matplotlib/sns/plotly/pandas и набор данных представленный ранее визуализируйте

````python
import matplotlib.pyplot as plt
````

1. Как менялась средняя цена квартиры в Нагатинском Затоне(Nagatinski Zaton) с течением времени

````python
df['timestamp'] = pd.to_datetime(df['timestamp'])
df['timestamp']
````

````python
nag_df = df[df['sub_area'] == 'Nagatinskij Zaton'].copy()
mean_month_prices = nag_df.groupby(nag_df['timestamp'].dt.to_period('M'))['price_doc'].mean()
mean_month_prices /= 1000000

plt.figure(figsize=(10, 5))
plt.plot(mean_month_prices.index.to_timestamp(), mean_month_prices, marker='o')
plt.title('Средняя цена квартиры в Нагатинском Затоне (млн руб.)')
plt.xlabel('Дата')
plt.ylabel('Млн руб.')
plt.grid(True)
plt.show()
````

2. Постройте гистограмму распределения цен на квартиры с общей площадью более 50 кв.метров

````python
cond = df[df['full_sq'] > 50]
cond = cond['price_doc'] / 1000000

plt.figure(figsize=(10, 5))
plt.hist(cond, bins=50, edgecolor='black')

plt.title('Распределение цен на квартиры площадью более 50 кв. м')
plt.xlabel('Цена (млн руб.)')
plt.ylabel('Количество квартир')
plt.grid(True, alpha=0.5)

plt.show()
````

3. Визуализируйте количество проданных квартир в разных районах

````python
top_areas = df['sub_area'].value_counts().head(25).sort_values()

plt.figure(figsize=(10, 6))
top_areas.plot(kind='barh', color='#1f77b4', edgecolor='black')

plt.title('Топ-25 районов по количеству проданных квартир')
plt.xlabel('Количество сделок')
plt.ylabel('Район')
plt.grid(axis='x', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
````

4. Постройте box plot отображающий распределение полных площадей квартир в Басманном районе

````python
basman_sq = df[df['sub_area'] == 'Basmannoe']

plt.figure(figsize=(6, 8))
plt.boxplot(basman_sq['full_sq'], patch_artist=True)

plt.title('Распределение полной площади квартир в Басманном районе')
plt.ylabel('Полная площадь (кв. м)')
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
````

5. Визуализируйте зависимость средней стоимости квартир в зависимости от этажа

````python
price_for_floor = df.groupby('floor')['price_doc'].mean()
price_for_floor /= 1000000

plt.figure(figsize=(12, 5))
plt.plot(price_for_floor.index, price_for_floor.values, marker = 'o', color='purple')
plt.title('Средняя стоимость квартир по этажам')
plt.xlabel('Этаж')
plt.ylabel('Средняя цена (млн руб.)')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
````

6. Покажите как меняется средняя площадь квартир в зависимости от района

````python
sq_for_area = df.groupby('sub_area')['full_sq'].mean().sort_values()
mid_start = len(sq_for_area) // 2 - 5

top10 = sq_for_area.head(10)
mid10 = sq_for_area[mid_start:].head(10)
end10 = sq_for_area.tail(10)

fig, axes = plt.subplots(1, 3, figsize=(18, 6), sharex=True)

axes[0].barh(top10.index, top10.values, color='green', edgecolor='black')
axes[0].set_title('Первые 10 районов (наименьшая площадь)')
axes[0].set_xlabel('Площадь (кв. м)')
axes[0].grid(axis='x', linestyle='--', alpha=0.5)

axes[1].barh(mid10.index, mid10.values, color='yellow', edgecolor='black')
axes[1].set_title('Средние 10 районов')
axes[1].set_xlabel('Площадь (кв. м)')
axes[1].grid(axis='x', linestyle='--', alpha=0.5)

axes[2].barh(end10.index, end10.values, color='red', edgecolor='black')
axes[2].set_title('Последние 10 районов (наибольшая площадь)')
axes[2].set_xlabel('Площадь (кв. м)')
axes[2].grid(axis='x', linestyle='--', alpha=0.5)

plt.suptitle('Изменение средней площади квартир в зависимости от районов', fontsize=14, y=1.02)
plt.tight_layout()
plt.show()
````

7. Постройте график изменения медианной цены на квартиру менее 50 кв. метров за период с 2011 до 2014 года*

\* Задание на доп. балл

````python
new_df = df[(df['full_sq'] < 50) & (df['timestamp'] >= '2011-01-01') & (df['timestamp'] < '2015-01-01')].copy()

grouped = new_df.groupby(new_df['timestamp'].dt.to_period('M'))['price_doc'].median()
grouped /= 1000000

plt.figure(figsize=(10, 5))
plt.plot(grouped.index.to_timestamp(), grouped, marker='o', color="red")
plt.title('Медиаонная цена квартиры меньше 50 кв. м. за 2011-2014 гг. (млн руб.)')
plt.xlabel('Дата')
plt.ylabel('Млн руб.')
plt.grid(True)
plt.show()
````
